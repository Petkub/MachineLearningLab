## Problem 8 — find the bugs

The code tries to:
- filter Titanic to **female passengers**
- compute their **mean fare**
- plot a **bar chart** of mean fare per `Pclass`

It runs but the output is wrong. Find and fix **three bugs**.

````python
rng = np.random.default_rng(1)
df = pd.DataFrame({
    "Sex":    rng.choice(["female", "male"], size=200),
    "Pclass": rng.integers(1, 4, size=200),
    "Fare":   rng.uniform(5, 100, size=200),
})

# BROKEN — fix three things
females      = df[df["Sex"] == "Male"]              # bug 1
mean_fare    = females["Fare"].sum()                # bug 2
fares_by_cls = df.groupby("Pclass")["Fare"].max()   # bug 3

plt.bar(fares_by_cls.index.astype(str), fares_by_cls.values)
plt.title("Mean Fare per Class")
plt.show()

# save your fixed values:
fixed_females      = ...
fixed_mean_fare    = ...
fixed_fares_by_cls = ...
````

````python
# verify against ground truth
exp_females      = df[df["Sex"] == "female"]
exp_mean         = exp_females["Fare"].mean()
exp_fares_by_cls = df.groupby("Pclass")["Fare"].mean()

assert len(fixed_females) == len(exp_females), "wrong filter"
assert abs(fixed_mean_fare - exp_mean) < 1e-6, "wrong aggregation"
assert fixed_fares_by_cls.equals(exp_fares_by_cls), "wrong group aggregation"
print("Q8 ok")
````

<details><summary>Hint</summary>

Three bugs:
1. Filter: `"Male"` (capital M) ≠ values in column (lowercase). Compare with `"female"`.
2. `.sum()` gives total — you want MEAN.
3. `.max()` gives biggest fare per class — you want MEAN.
