```python
# Q1 — hist
plt.figure(figsize=(8, 4))
plt.hist(df['Age'], bins=30)
plt.xlabel('Age'); plt.ylabel('Count'); plt.title('Age distribution')
plt.show()

# Q2 — bar
counts = df['Pclass'].value_counts()
ax = counts.plot.bar(figsize=(6, 4))
for i, c in enumerate(counts):
    plt.text(i, c + 5, str(c), ha='center', fontweight='bold')
plt.title('Pclass counts'); plt.ylabel('count')
plt.show()

# Q3 — scatter colored
plt.figure(figsize=(8, 5))
plt.scatter(df['Age'], df['Fare'], c=df['Survived'], alpha=0.5, cmap='coolwarm')
plt.xlabel('Age'); plt.ylabel('Fare')
plt.colorbar(label='Survived')
plt.show()

# Q4 — boxplot per group
df.boxplot(column='Fare', by='Pclass', figsize=(8, 5))
plt.suptitle('')
plt.title('Fare by Pclass')
plt.show()

# Q5 — subplots
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes[0, 0].hist(df['Age'], bins=30); axes[0, 0].set_title('Age')
axes[0, 1].hist(df['Fare'], bins=30); axes[0, 1].set_title('Fare')
df['Pclass'].value_counts().plot.bar(ax=axes[1, 0]); axes[1, 0].set_title('Pclass')
axes[1, 1].scatter(df['Age'], df['Fare'], alpha=0.4); axes[1, 1].set_title('Age vs Fare')
plt.tight_layout(); plt.show()

# Q6 — fixes
plt.plot(epochs, train_loss, label='train')   # x then y
plt.plot(epochs, val_loss, label='val')       # added label
plt.xlabel('epoch'); plt.ylabel('loss')
plt.title('Training curve')
plt.legend(); plt.show()
```
