## Part 2 — Dictionaries

A dictionary looks things up **by name** instead of by position.
Curly braces, and every entry is `key: value`.

````python
prices = {"pad thai": 60, "green curry": 80, "mango sticky rice": 45}

print(prices["pad thai"])         # look up by key
prices["tom yum"] = 70            # new key  -> adds it
prices["pad thai"] = 65           # key exists -> overwrites it
print(prices)
print("pizza" in prices)          # `in` checks the KEYS, never the values
````

## Part 3 — List of dictionaries

Real data is **many records, each with named fields**:
rows are the list, columns are the keys.

`students[1]["score"]` — index first (picks the row), key second (picks the field).

````python
students = [
    {"name": "Ann", "score": 82, "club": "robotics"},
    {"name": "Ben", "score": 91, "club": "music"},
    {"name": "Cat", "score": 67, "club": "robotics"},
    {"name": "Dan", "score": 74, "club": "art"},
    {"name": "Eve", "score": 88, "club": "music"},
]

print(students[0])              # one whole record — a dictionary
print(students[0]["name"])      # one field
print(len(students), "students")

for s in students:              # each turn, s is one student
    print(s["name"], s["score"])
````
