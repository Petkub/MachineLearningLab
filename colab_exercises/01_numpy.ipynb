## Problem 07 — find the bugs in this normalizer

The cell below tries to **min-max normalize** an array (rescale every element so the min becomes 0 and the max becomes 1). It runs but the output is wrong. Find and fix **three bugs**.

Things to look for: wrong axis, integer division, accidentally using a Python loop, wrong shape returned by an aggregation.

````python
# BROKEN — fix three things, save the corrected normalized array in `norm_q7`.
#
# Bugs to find:
#  - using mean instead of min/max — rescale needs the extremes
#  - axis is wrong, so each row uses the wrong reference values
#  - the formula is missing one operand, so the result isn't in [0, 1]

raw = np.array([[1.0, 5.0, 9.0],
                [2.0, 4.0, 6.0],
                [10.0, 20.0, 30.0]])

row_mean = raw.mean(axis=0)
row_max  = raw.max(axis=0)
norm_q7  = (raw - row_mean) / row_max

print('norm_q7:\n', norm_q7)
````

````python
# correct min-max per row: (x - row_min) / (row_max - row_min)
expected = (raw - raw.min(axis=1, keepdims=True)) / (raw.max(axis=1, keepdims=True) - raw.min(axis=1, keepdims=True))
assert np.allclose(norm_q7, expected), f'still buggy — your output:\n{norm_q7}\nexpected:\n{expected}'
assert np.allclose(norm_q7.min(axis=1), 0)
assert np.allclose(norm_q7.max(axis=1), 1)
print('Q7 ok — every row now spans [0, 1]')
````

<details><summary>Hint</summary>

Three independent failures:
- Min-max rescaling needs `min` and `max`, not `mean`. The `mean` produces the wrong reference point.
- Per-row aggregation collapses the **column** dimension → that's `axis=1`, not `axis=0`. Otherwise every row uses column-wise stats and broadcasts wrong.
- The formula is `(x - min) / (max - min)`, not `(x - min) / max`. The denominator is the **range**, not just the maximum.

When subtracting a per-row vector from a 2-D array, you may also need `keepdims=True` to keep the broadcast shape sane.

</details>
