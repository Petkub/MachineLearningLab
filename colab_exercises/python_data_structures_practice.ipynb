**How this one works**

Every task asks you to finish a **function**. The `CHECK` cell below it runs your
function on **several different test cases** — not just one. That matters:
code that works on one example is often still broken.

```
def double_all(nums):
    # TODO
    return []
```

Run the CHECK. If every case passes you get `ALL 5 CASES PASS`.
If one fails you are told **which input** broke it and what you returned.

Read the failing input carefully — the empty list and the one-item list
catch more bugs than anything else.

````python
# Run this first. It is the little test runner every CHECK cell uses.

def run_cases(fn, cases):
    passed = 0
    for i, (args, expected) in enumerate(cases, 1):
        try:
            got = fn(*[a.copy() if isinstance(a, (list, dict)) else a for a in args])
        except Exception as e:
            print(f"CASE {i}  CRASHED   input={args}  ->  {type(e).__name__}: {e}")
            continue
        if got == expected:
            passed += 1
        else:
            print(f"CASE {i}  WRONG     input={args}\n         expected {expected!r}\n         got      {got!r}")
    if passed == len(cases):
        print(f"ALL {passed} CASES PASS")
    else:
        print(f"{passed}/{len(cases)} passed — keep going")

print("test runner ready")
````
